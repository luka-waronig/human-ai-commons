Sobol sensitivity analysis uses one replicate mean per sampling-design row. Run the notebook in the pinned Python 3.12 environment. Model equations, estimator definitions, and execution commands are described in `docs/model-and-analysis.md`.


```python
from sensitivity_analysis import WaterToCeSensitivityAnalyzer
analyzer = WaterToCeSensitivityAnalyzer(n_runs_per_config=4, max_steps=100, n_jobs=1, seed=42)
print(analyzer.param_bounds)
```

## One-at-a-time effects

Compare each output in its own units. The plotted bands show standard deviations across replications, not confidence intervals.


```python
oat = analyzer.oat_sensitivity_analysis(n_samples=8)
analyzer.plot_oat_results(oat)
```

## Sobol indices of the replicate-mean response

Use a power-of-two base design size. There is one conditional mean per design row, in the original order. Finite-replication noise can still bias the indices; inspect within-configuration variance and repeat with more replications. A constant output has undefined variance-based indices.


```python
sobol = analyzer.sobol_sensitivity_analysis(n_samples=128)
print(sobol['within_config_variances'].mean())
analyzer.plot_sobol_results(sobol)
```

## Descriptive distribution sensitivity

Compare the common background ensemble with each input fixed to its baseline. Report a KS distance between configuration means; the paired design does not justify the ordinary independent-sample KS p-value.


```python
ks = analyzer.ks_sensitivity_analysis(n_samples=64)
analyzer.plot_ks_results(ks)
```
