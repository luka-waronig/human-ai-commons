Sobol sensitivity analysis uses one replicate mean per sampling-design row. The numerical environment is pinned to Python 3.12. Model equations, estimator definitions and execution commands are described in `docs/model-and-analysis.md`.


```python
from sensitivity_analysis import WaterToCeSensitivityAnalyzer
analyzer = WaterToCeSensitivityAnalyzer(n_runs_per_config=4, max_steps=100, n_jobs=1, seed=42)
print(analyzer.param_bounds)
```

## One-at-a-time effects

Each output is plotted in its own units. Bands show standard deviations across replications, not confidence intervals.


```python
oat = analyzer.oat_sensitivity_analysis(n_samples=8)
analyzer.plot_oat_results(oat)
```

## Sobol indices of the replicate-mean response

The base design size is a power of two. Each sampling-design row contributes one conditional mean, in the original order. Finite-replication noise can bias the indices; within-configuration variance records that uncertainty. A constant output has undefined variance-based indices.


```python
sobol = analyzer.sobol_sensitivity_analysis(n_samples=128)
print(sobol['within_config_variances'].mean())
analyzer.plot_sobol_results(sobol)
```

## Descriptive distribution sensitivity

The KS distance compares configuration means in a common background ensemble with each input fixed to its baseline. The paired design supplies a descriptive distance rather than an ordinary independent-sample KS p-value.


```python
ks = analyzer.ks_sensitivity_analysis(n_samples=64)
analyzer.plot_ks_results(ks)
```

## Per-output report

Output ranges retain their physical units. Dimensionless sensitivity indices are reported separately.


```python
report = analyzer.generate_sensitivity_report(oat, sobol, ks)
report
```
